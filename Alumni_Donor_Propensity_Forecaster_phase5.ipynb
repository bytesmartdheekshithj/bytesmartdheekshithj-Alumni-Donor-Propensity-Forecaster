```python
# COMMAND: Train the real 2025 Alumni Donor Propensity model with the corrected feature types

import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# Load dataset
data = pd.read_csv("alumni_donor_dataset_2025.csv")

print("Dataset shape:", data.shape)

# Numerical features
NUMERICAL_FEATURES = [
    "Age",
    "Graduation_Year",
    "Event_Attendance",
    "Email_Open_Rate",
    "Consecutive_Giving_Years",
    "Donation_Last_Year_2024",
    "Total_Lifetime_Giving"
]

# Categorical features
# Wealth_Rating is included here because it contains values such as "Unknown"
CATEGORICAL_FEATURES = [
    "Gender",
    "Degree_Level",
    "Major",
    "Alumni_Status",
    "Email_Available",
    "Phone_Available",
    "Wealth_Rating",
    "Income_Bracket",
    "Location"
]

TARGET = "Is_Donor_2025"

# Prepare features and target
X = data[NUMERICAL_FEATURES + CATEGORICAL_FEATURES].copy()

y = data[TARGET].map({
    "Yes": 1,
    "No": 0
})

print("\nTarget distribution:")
print(y.value_counts())

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, NUMERICAL_FEATURES),
    ("categorical", categorical_pipeline, CATEGORICAL_FEATURES)
])

# Logistic Regression model
model_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train
model_pipeline.fit(X_train, y_train)

print("\nModel training completed successfully!")

# Save the REAL trained model
MODEL_PATH = "alumni_donor_propensity_model_2025.pkl"

joblib.dump(model_pipeline, MODEL_PATH)

print("Real trained model saved successfully!")
print("Model file:", MODEL_PATH)
```

```python
# COMMAND: Create a fresh Flask API with health-check and prediction endpoints

from flask import Flask, request, jsonify
from flask_cors import CORS

import pandas as pd
import joblib
import json
import redis
import pika
import threading
import time
import socket # Import socket to find available ports

# --------------------------------------------------
# Create a completely new Flask application
# --------------------------------------------------

new_app = Flask(__name__)
CORS(new_app)

# --------------------------------------------------
# Load ML model
# --------------------------------------------------

MODEL_PATH = "alumni_donor_propensity_model_2025.pkl"

model = joblib.load(MODEL_PATH)

print("ML model loaded successfully")
print("Model type:", type(model))

# --------------------------------------------------
# Redis configuration
# --------------------------------------------------

REDIS_HOST = "129.153.75.221"
REDIS_PORT = 6379
REDIS_USERNAME = "default"
REDIS_PASSWORD = "cwe5cU6Tyzvd"

redis_client = redis.Redis(
    host=REDIS_HOST,
    port=REDIS_PORT,
    username=REDIS_USERNAME,
    password=REDIS_PASSWORD,
    decode_responses=True,
    socket_connect_timeout=10
)

try:
    redis_client.ping()
    print("Redis connected successfully")
except Exception as e:
    print("Redis connection failed:", e)

# --------------------------------------------------
# RabbitMQ configuration
# --------------------------------------------------

RABBITMQ_HOST = "129.153.75.221"
RABBITMQ_PORT = 5672
RABBITMQ_USERNAME = "bytesmart_interns"
RABBITMQ_PASSWORD = "YaZU4ghFdBzY"

QUEUE_NAME = "alumni_donation_queue"

rabbit_connection = None
rabbit_channel = None

try:

    credentials = pika.PlainCredentials(
        RABBITMQ_USERNAME,
        RABBITMQ_PASSWORD
    )

    rabbit_connection = pika.BlockingConnection(
        pika.ConnectionParameters(
            host=RABBITMQ_HOST,
            port=RABBITMQ_PORT,
            credentials=credentials
        )
    )

    rabbit_channel = rabbit_connection.channel()

    rabbit_channel.queue_declare(
        queue=QUEUE_NAME,
        durable=True
    )

    print("RabbitMQ connected successfully")
    print("Queue:", QUEUE_NAME)

except Exception as e:

    print("RabbitMQ connection failed:", e)

# --------------------------------------------------
# Prediction function
# --------------------------------------------------

def generate_prediction(alumni_data):

    model_input = pd.DataFrame([alumni_data])

    # Alumni_ID is only an identifier.
    # It must not be used as an ML feature.
    model_input = model_input.drop(
        columns=["Alumni_ID"],
        errors="ignore"
    )

    probability = model.predict_proba(
        model_input
    )[0][1]

    propensity_score = round(
        float(probability * 100),
        2
    )

    if propensity_score >= 80:

        category = "High"
        interpretation = "Likely to Donate"

    elif propensity_score >= 50:

        category = "Medium"
        interpretation = "Moderate Donation Likelihood"

    else:

        category = "Low"
        interpretation = "Lower Donation Likelihood"

    return {
        "Alumni_ID": alumni_data["Alumni_ID"],
        "donation_probability": round(
            float(probability),
            4
        ),
        "propensity_score": propensity_score,
        "category": category,
        "interpretation": interpretation
    }

# --------------------------------------------------
# Health-check route
# --------------------------------------------------

@new_app.route("/", methods=["GET"])
def home():

    return jsonify({
        "status": "running",
        "project": "Alumni Donor Propensity Forecaster",
        "model": "2025",
        "redis": "connected",
        "rabbitmq": "connected"
    })

# --------------------------------------------------
# Prediction route
# --------------------------------------------------

@new_app.route("/predict", methods=["POST"])
def predict():

    print("\nPrediction request received")

    data = request.get_json()

    if not data:

        return jsonify({
            "error": "Request body is empty"
        }), 400

    if "Alumni_ID" not in data:

        return jsonify({
            "error": "Alumni_ID is required"
        }), 400

    alumni_id = data["Alumni_ID"]

    cache_key = f"alumni:{alumni_id}"

    # --------------------------------------------------
    # Check Redis cache
    # --------------------------------------------------

    try:

        cached_result = redis_client.get(cache_key)

    except Exception as e:

        print("Redis cache error:", e)
        cached_result = None

    if cached_result:

        print("Redis Cache HIT")

        return jsonify(
            json.loads(cached_result)
        )

    print("Redis Cache MISS")

    # --------------------------------------------------
    # Generate ML prediction
    # --------------------------------------------------

    try:

        result = generate_prediction(data)

        print("ML prediction generated")

    except Exception as e:

        print("Prediction error:", str(e))

        return jsonify({
            "error": str(e)
        }), 400

    # --------------------------------------------------
    # Store result in Redis
# --------------------------------------------------

    try:

        redis_client.set(
            cache_key,
            json.dumps(result),
            ex=3600
        )

        print("Prediction stored in Redis")

    except Exception as e:

        print("Redis storage failed:", e)

    # --------------------------------------------------
    # Publish result to RabbitMQ
    # --------------------------------------------------

    if rabbit_channel is not None:

        try:

            rabbit_channel.basic_publish(
                exchange="",
                routing_key=QUEUE_NAME,
                body=json.dumps(result),
                properties=pika.BasicProperties(
                    delivery_mode=2
                )
            )

            print("Prediction published to RabbitMQ")

        except Exception as e:

            print("RabbitMQ publishing failed:", e)

    # --------------------------------------------------
    # Return prediction
    # --------------------------------------------------

    return jsonify(result)

# --------------------------------------------------
# Function to find an available port
# --------------------------------------------------
def find_available_port(start_port, end_port):
    for port in range(start_port, end_port + 1):
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
            try:
                s.bind(("0.0.0.0", port))
                return port
            except OSError:
                continue
    raise RuntimeError(f"No available ports in range {start_port}-{end_port}")

# --------------------------------------------------
# Start Flask in background
# --------------------------------------------------
def run_flask_on_port(port):
    new_app.run(
        host="0.0.0.0",
        port=port,
        debug=False,
        use_reloader=False
    )

try:
    # Find an available port, starting from 5000 and checking up to 5010
    current_port = find_available_port(5000, 5010)
    print(f"Starting Flask API on port {current_port}")

    flask_thread = threading.Thread(
        target=run_flask_on_port,
        args=(current_port,),
        daemon=True
    )
    flask_thread.start()

    time.sleep(3)

    print(f"\nFlask API started successfully on port {current_port}")
    print(f"Health endpoint: http://127.0.0.1:{current_port}/")
    print(f"Prediction endpoint: http://127.0.0.1:{current_port}/predict")

except RuntimeError as e:
    print(f"Failed to start Flask API: {e}")
    print("Please restart the Colab runtime if the issue persists or try a different port range.")
```

```python
# COMMAND: Create the complete Alumni Donor Propensity Flask backend

from flask import Flask, request, jsonify
from flask_cors import CORS

import pandas as pd
import joblib
import json
import redis
import pika
import threading
import time

# ============================================================
# 1. Flask application
# ============================================================

app = Flask(__name__)
CORS(app)

PORT = 5050

# ============================================================
# 2. ML MODEL
# ============================================================

MODEL_PATH = "alumni_donor_propensity_model_2025.pkl"

model = joblib.load(MODEL_PATH)

print("ML model loaded successfully")
print("Model type:", type(model))

# ============================================================
# 3. REDIS CONFIGURATION
# ============================================================

REDIS_HOST = "129.153.75.221"
REDIS_PORT = 6379
REDIS_USERNAME = "default"

# Enter your working Redis password here
REDIS_PASSWORD = "cwe5cU6Tyzvd"

redis_client = redis.Redis(
    host=REDIS_HOST,
    port=REDIS_PORT,
    username=REDIS_USERNAME,
    password=REDIS_PASSWORD,
    decode_responses=True,
    socket_connect_timeout=10
)

try:
    redis_client.ping()
    print("Redis connected successfully")
except Exception as e:
    print("Redis connection failed:", e)

# ============================================================
# 4. RABBITMQ CONFIGURATION
# ============================================================

RABBITMQ_HOST = "129.153.75.221"
RABBITMQ_PORT = 5672

# Enter your working RabbitMQ credentials here
RABBITMQ_USERNAME = "bytesmart_interns"
RABBITMQ_PASSWORD = "YaZU4ghFdBzY"

QUEUE_NAME = "alumni_donation_queue"

# ============================================================
# 5. RabbitMQ connection function
# ============================================================

def create_rabbitmq_connection():

    credentials = pika.PlainCredentials(
        RABBITMQ_USERNAME,
        RABBITMQ_PASSWORD
    )

    connection = pika.BlockingConnection(
        pika.ConnectionParameters(
            host=RABBITMQ_HOST,
            port=RABBITMQ_PORT,
            credentials=credentials,
            heartbeat=60,
            blocked_connection_timeout=30
        )
    )

    channel = connection.channel()

    channel.queue_declare(
        queue=QUEUE_NAME,
        durable=True
    )

    return connection, channel


# ============================================================
# 6. Test RabbitMQ connection
# ============================================================

rabbit_connection = None
rabbit_channel = None

try:

    rabbit_connection, rabbit_channel = create_rabbitmq_connection()

    print("RabbitMQ connected successfully")
    print("Queue:", QUEUE_NAME)

except Exception as e:

    print("RabbitMQ connection failed:")
    print(type(e).__name__, e)


# ============================================================
# 7. Generate ML prediction
# ============================================================

def generate_prediction(alumni_data):

    model_input = pd.DataFrame([alumni_data])

    # Alumni_ID is only an identifier.
    # It must NOT be used as an ML feature.
    model_input = model_input.drop(
        columns=["Alumni_ID"],
        errors="ignore"
    )

    probability = model.predict_proba(
        model_input
    )[0][1]

    propensity_score = round(
        float(probability * 100),
        2
    )

    if propensity_score >= 80:

        category = "High"
        interpretation = "Likely to Donate"

    elif propensity_score >= 50:

        category = "Medium"
        interpretation = "Moderate Donation Likelihood"

    else:

        category = "Low"
        interpretation = "Lower Donation Likelihood"

    return {
        "Alumni_ID": alumni_data["Alumni_ID"],
        "donation_probability": round(
            float(probability),
            4
        ),
        "propensity_score": propensity_score,
        "category": category,
        "interpretation": interpretation
    }


# ============================================================
# 8. RabbitMQ publish function
# ============================================================

def publish_to_rabbitmq(result):

    global rabbit_connection
    global rabbit_channel

    try:

        # Check existing connection
        if (
            rabbit_connection is None
            or rabbit_connection.is_closed
        ):

            print("RabbitMQ connection closed. Reconnecting...")

            rabbit_connection, rabbit_channel = (
                create_rabbitmq_connection()
            )

        message = json.dumps(result)

        rabbit_channel.basic_publish(
            exchange="",
            routing_key=QUEUE_NAME,
            body=message,
            properties=pika.BasicProperties(
                delivery_mode=2,
                content_type="application/json"
            )
        )

        print("Prediction published to RabbitMQ")

        return True

    except Exception as e:

        print("RabbitMQ publish failed:")
        print(type(e).__name__, e)

        # Try one reconnect
        try:

            print("Trying RabbitMQ reconnect...")

            rabbit_connection, rabbit_channel = (
                create_rabbitmq_connection()
            )

            rabbit_channel.basic_publish(
                exchange="",
                routing_key=QUEUE_NAME,
                body=json.dumps(result),
                properties=pika.BasicProperties(
                    delivery_mode=2,
                    content_type="application/json"
                )
            )

            print("RabbitMQ publish successful after reconnect")

            return True

        except Exception as retry_error:

            print("RabbitMQ retry failed:")
            print(type(retry_error).__name__, retry_error)

            return False


# ============================================================
# 9. Health endpoint
# ============================================================

@app.route("/", methods=["GET"])
def home():

    return jsonify({
        "status": "running",
        "project": "Alumni Donor Propensity Forecaster",
        "model": "2025",
        "redis": "connected",
        "rabbitmq": "connected",
        "port": PORT
    })


# ============================================================
# 10. Prediction endpoint
# ============================================================

@app.route("/predict", methods=["POST"])
def predict():

    print("\nPrediction request received")

    data = request.get_json()

    if not data:

        return jsonify({
            "error": "Request body is empty"
        }), 400

    if "Alumni_ID" not in data:

        return jsonify({
            "error": "Alumni_ID is required"
        }), 400

    alumni_id = data["Alumni_ID"]

    cache_key = f"alumni:{alumni_id}"

    # ========================================================
    # Redis Cache Check
    # ========================================================

    try:

        cached_result = redis_client.get(
            cache_key
        )

    except Exception as e:

        print("Redis read failed:", e)

        cached_result = None

    if cached_result:

        print("Redis Cache HIT")

        return jsonify(
            json.loads(cached_result)
        )

    print("Redis Cache MISS")

    # ========================================================
    # ML Prediction
    # ========================================================

    try:

        result = generate_prediction(data)

        print("ML prediction generated")

    except Exception as e:

        print("Prediction error:", e)

        return jsonify({
            "error": str(e)
        }), 400

    # ========================================================
    # Store in Redis
    # ========================================================

    try:

        redis_client.set(
            cache_key,
            json.dumps(result),
            ex=3600
        )

        print("Prediction stored in Redis")

    except Exception as e:

        print("Redis storage failed:", e)

    # ========================================================
    # RabbitMQ
    # ========================================================

    rabbit_status = publish_to_rabbitmq(result)

    # ========================================================
    # API response
    # ========================================================

    result["rabbitmq_published"] = rabbit_status

    return jsonify(result)


# ============================================================
# 11. Start Flask
# ============================================================

def run_flask():

    app.run(
        host="0.0.0.0",
        port=PORT,
        debug=False,
        use_reloader=False
    )


flask_thread = threading.Thread(
    target=run_flask,
    daemon=True
)

flask_thread.start()

time.sleep(3)

print("\nFlask API started")
print("Health URL:")
print(f"http://127.0.0.1:{PORT}/")

print("\nPrediction URL:")
print(f"http://127.0.0.1:{PORT}/predict")
```

```python
# COMMAND: Receive and acknowledge a prediction from RabbitMQ

import pika
import json

credentials = pika.PlainCredentials(
    RABBITMQ_USERNAME,
    RABBITMQ_PASSWORD
)

consumer_connection = pika.BlockingConnection(
    pika.ConnectionParameters(
        host=RABBITMQ_HOST,
        port=RABBITMQ_PORT,
        credentials=credentials,
        heartbeat=60
    )
)

consumer_channel = consumer_connection.channel()

consumer_channel.queue_declare(
    queue=QUEUE_NAME,
    durable=True
)

method_frame, header_frame, body = (
    consumer_channel.basic_get(
        queue=QUEUE_NAME,
        auto_ack=False
    )
)

if method_frame:

    received_message = json.loads(
        body.decode("utf-8")
    )

    print("Message received successfully")

    print(
        json.dumps(
            received_message,
            indent=4
        )
    )

    consumer_channel.basic_ack(
        delivery_tag=method_frame.delivery_tag
    )

    print("\nMessage acknowledged successfully")

else:

    print("No message available in RabbitMQ queue")

consumer_connection.close()
```

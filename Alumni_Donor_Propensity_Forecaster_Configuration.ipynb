```python
# COMMAND: Define project and configuration paths

import os

project_path = "/content/drive/MyDrive/Alumni_Donor_Propensity_Forecaster"

config_path = os.path.join(project_path, "config")
model_path = os.path.join(
    project_path,
    "models",
    "alumni_donor_model_pipeline.pkl"
)

os.makedirs(config_path, exist_ok=True)

print("Project path:", project_path)
print("Configuration path:", config_path)
print("Model path:", model_path)

print("\nModel exists:", os.path.exists(model_path))
```

```python
# COMMAND: Create private environment configuration file

env_content = """PORT=5000
MODEL_PATH=/content/drive/MyDrive/Alumni_Donor_Propensity_Forecaster/models/alumni_donor_model_pipeline.pkl
CACHE_ENABLED=True
HIGH_THRESHOLD=70
MEDIUM_THRESHOLD=40
"""

env_path = os.path.join(config_path, ".env")

with open(env_path, "w") as f:
    f.write(env_content)

print("Private .env configuration file created successfully!")
print("Location:", env_path)
```

```python
# COMMAND: Create safe environment configuration template

env_example_content = """PORT=5000
MODEL_PATH=/content/drive/MyDrive/Alumni_Donor_Propensity_Forecaster/models/alumni_donor_model_pipeline.pkl
CACHE_ENABLED=True
HIGH_THRESHOLD=70
MEDIUM_THRESHOLD=40

# Add credentials only as placeholders
# API_KEY=your_api_key_here
# NGROK_AUTH_TOKEN=your_ngrok_token_here
"""

env_example_path = os.path.join(config_path, ".env.example")

with open(env_example_path, "w") as f:
    f.write(env_example_content)

print("Safe .env.example created successfully!")
print("Location:", env_example_path)
```

```python
# COMMAND: Create centralized configuration module

config_code = '''import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

class Config:

    # Flask/API configuration
    PORT = int(os.getenv("PORT", "5000"))

    # Machine learning model configuration
    MODEL_PATH = os.getenv(
        "MODEL_PATH",
        "/content/drive/MyDrive/Alumni_Donor_Propensity_Forecaster/models/alumni_donor_model_pipeline.pkl"
    )

    # Cache configuration
    CACHE_ENABLED = os.getenv(
        "CACHE_ENABLED",
        "True"
    ).lower() == "true"

    # Donation propensity thresholds
    HIGH_THRESHOLD = float(
        os.getenv("HIGH_THRESHOLD", "70")
    )

    MEDIUM_THRESHOLD = float(
        os.getenv("MEDIUM_THRESHOLD", "40")
    )
'''

config_file = os.path.join(config_path, "config.py")

with open(config_file, "w") as f:
    f.write(config_code)

print("config.py created successfully!")
print("Location:", config_file)
```

```python
# COMMAND: Load and verify configuration

import sys

if config_path not in sys.path:
    sys.path.append(config_path)

from config import Config

print("Configuration loaded successfully!")

print("\nConfiguration values:")
print("-----------------------------")
print("PORT:", Config.PORT)
print("MODEL_PATH:", Config.MODEL_PATH)
print("CACHE_ENABLED:", Config.CACHE_ENABLED)
print("HIGH_THRESHOLD:", Config.HIGH_THRESHOLD)
print("MEDIUM_THRESHOLD:", Config.MEDIUM_THRESHOLD)
```

```python
# COMMAND: Verify configuration files exist

files_to_check = [
    env_path,
    env_example_path,
    config_file
]

for file_path in files_to_check:
    print(
        os.path.basename(file_path),
        "->",
        "Exists" if os.path.exists(file_path) else "Missing"
    )
```

```python
# COMMAND: Verify trained model using configured model path

import joblib

print("Loading model from configuration...")

model = joblib.load(Config.MODEL_PATH)

print("\nModel loaded successfully!")
print("Model type:", type(model))
print("Has predict():", hasattr(model, "predict"))
print("Has predict_proba():", hasattr(model, "predict_proba"))
```

```python
# COMMAND: Verify trained model pipeline components

if hasattr(model, "named_steps"):

    print("Pipeline components:")

    for name, step in model.named_steps.items():
        print(f"{name} -> {type(step)}")

else:
    print("Model does not contain named pipeline steps.")
```

```python
# COMMAND: Test High Medium Low propensity thresholds

test_scores = [85, 55, 25]

for propensity_score in test_scores:

    if propensity_score >= Config.HIGH_THRESHOLD:
        category = "High"

    elif propensity_score >= Config.MEDIUM_THRESHOLD:
        category = "Medium"

    else:
        category = "Low"

    print(
        f"Score: {propensity_score} -> "
        f"Category: {category}"
    )
```

```python
# COMMAND: Verify cache configuration

if Config.CACHE_ENABLED:
    print("Cache status: ENABLED")
else:
    print("Cache status: DISABLED")
```

```python
# COMMAND: Perform configuration security checks

print("Security checks")
print("-----------------------------")

print(
    ".env exists:",
    os.path.exists(env_path)
)

print(
    ".env.example exists:",
    os.path.exists(env_example_path)
)

print(
    "config.py exists:",
    os.path.exists(config_file)
)

print("\nPrivate .env contents are NOT displayed.")
print("Use .env.example for safe repository sharing.")
```

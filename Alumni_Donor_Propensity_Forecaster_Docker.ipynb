```python
# COMMAND: Upload trained model file

from google.colab import files
import os

uploaded = files.upload()

print("\nUploaded files:")

for filename in uploaded.keys():
    print("-", filename)

# Find the uploaded .pkl file
pkl_files = [
    filename
    for filename in uploaded.keys()
    if filename.endswith(".pkl")
]

if len(pkl_files) == 0:
    raise FileNotFoundError(
        "No .pkl model file was uploaded."
    )

model_filename = pkl_files[0]

print("\nModel file selected:")
print(model_filename)
```

```python
# COMMAND: Inspect uploaded PKL model structure

import joblib

model_object = joblib.load(model_filename)

print("Loaded object type:")
print(type(model_object))

print("\nObject contents:")

if isinstance(model_object, dict):

    print("This PKL contains a dictionary.")

    print("\nDictionary keys:")

    for key in model_object.keys():
        print("-", key)

else:

    print("This PKL contains:")
    print(model_object)
```

```python
# COMMAND: Inspect dictionary contents

if isinstance(model_object, dict):

    for key, value in model_object.items():

        print("=" * 60)
        print("KEY:", key)
        print("TYPE:", type(value))

        if hasattr(value, "predict"):
            print("✓ This object has predict()")

        elif hasattr(value, "predict_proba"):
            print("✓ This object has predict_proba()")

        else:
            print("No predict method")

        print("=" * 60)
```
